#Mean, Median, Mode of Columns

```python
m_cols = ['Weekly_Study_Hours', 'Average_Sleep_Hours', 'Daily_Screen_Time_Hours', 'Stress_Score', 'Academic_Readiness_Score']

stats_data = []
for col in m_cols:
  mean_val = df[col].mean()
  median_val = df[col].median()
  mode_val = df[col].mode()[0]
  stats_data.append({'Column': col, 'Mean': mean_val, 'Median': median_val, 'Mode': mode_val})

stats_df = pd.DataFrame(stats_data)
display(stats_df)
```

##Formulae used
Mean = sum of observations / no. of observations

Median = middle value after sorting the observations

Mode = most frequently occuring observation value

#Range, Variance, Standard deviation, Q1, Q3, IQR

```python
from numpy._core.fromnumeric import var
measures_data = []
for col in m_cols:
  range_val = df[col].max() - df[col].min()
  var_val = df[col].var()
  std_val = df[col].std()
  q1_val = df[col].quantile(0.25)
  q3_val = df[col].quantile(0.75)
  iqr_val = q3_val - q1_val
  measures_data.append({'Column': col, 'Range': range_val, 'Variance': var_val, 'Standard Deviation': std_val, 'Q1': q1_val, 'Q3': q3_val, 'IQR': iqr_val})

measures_df = pd.DataFrame(measures_data)
display(measures_df)
```

```python
print("\nVariable with greatest variability:", df[m_cols].std().idxmax())
```

The variable having the greatest variability can be found by comparing the variance of different columns or finding the largest standard deviation.

In this case, by observation, it could be understood that the variable with the greatest variability is **Academic_Readiness_Score**

#Outlier Detection using IQR

```python
outlier_col = ['Weekly_Study_Hours', 'Daily_Screen_Time_Hours', 'Commute_Time_Minutes', 'Monthly_Discretionary_Spending']
```

```python
outlier_info = []
for col in outlier_col:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    outlier_info.append({
        'Column': col,
        'Q1': Q1,
        'Q3': Q3,
        'IQR': IQR,
        'Lower Bound': lower_bound,
        'Upper Bound': upper_bound,
        'Number of Outliers': len(outliers)
    })

outlier_df = pd.DataFrame(outlier_info)
print(outlier_df)
```

```python
outlier_stats = []

for col in outlier_col:
  mean_before = df[col].mean()
  median_before = df[col].median()

  q1 = df[col].quantile(0.25)
  q3 = df[col].quantile(0.75)
  iqr = q3 - q1
  low_bound = q1 - 1.5 * iqr
  high_bound = q3 + 1.5 * iqr

  df_filtered_outliers = df[(df[col] >= low_bound) & (df[col] <= high_bound)]

  mean_after = df_filtered_outliers[col].mean()
  median_after = df_filtered_outliers[col].median()

  outlier_stats.append({
      'Column': col,
      'Lower Bound': low_bound,
      'Upper Bound': high_bound,
      'Mean_Before_Outliers': mean_before,
      'Median_Before_Outliers': median_before,
      'Mean_After_Outliers': mean_after,
      'Median_After_Outliers': median_after
  })

outlier_comparison_df = pd.DataFrame(outlier_stats)
display(outlier_comparison_df)
```

#Probability

```python
A = df['Part_Time_Job'] == 'Yes'
B = df['Stress_Score'] >= 7
C = df['Scholarship'] == 'Yes'
D = df['Exercise_Days_Per_Week'] >= 3
```

```python
P_A = A.mean()
P_B = B.mean()
P_C = C.mean()
P_D = D.mean()

P_A_and_B = (A & B).mean()
P_A_or_B = P_A + P_B - P_A_and_B

P_A_given_B = P_A_and_B / P_B
P_B_given_A = P_A_and_B / P_A


print("P(A) =", P_A)
print("P(B) =", P_B)
print("P(C) =", P_C)
print("P(D) =", P_D)
print("\n")

print("P(A or B) =", P_A_or_B)
print("P(A and B) =", P_A_and_B)
print("\n")

print("P(A|B) =", P_A_given_B)
print("P(B|A) =", P_B_given_A)
```

##Formula
P(A) = no. of students satisfying condition A / total no. of students

P(A∪B) = P(A) + P(B) − P(A∩B)

P(A∩B) = no. of students satisfying both A and B / total students

P(A|B) = P(A∩B) / P(B)

#Mutually Exclusive Events

```python
yr_1 = df['Year_of_Study']==1
yr_4 = df['Year_of_Study']==4

P_yr1_and_yr4 = (yr_1 & yr_4).mean()
print(P_yr1_and_yr4)
```

The result of their intersection is 0, indicating that they are mutually exclusive.

A student can't be in Year 1 and Year 4 simultaneously, so they are mutually exclusive events

#Independent Events

Formula for independent events: P(A∩B)=P(A)×P(B)

```python
P_A_times_P_B = P_A * P_B

print("P(A and B) =", P_A_and_B)
print("P(A) × P(B) =", P_A_times_P_B)
print("Difference =", abs(P_A_and_B - P_A_times_P_B))
```

Since P(A and B) is approximately equal to P(A) × P(B), events A and B appear to be independent.

#Bayes' Theorem

P(A|B) = [P(B|A) * P(A)] / [(P(B|A) * P(A)) + (P(B|**~**A) * P(**~**A))]

```python
P_not_A = 1 - P_A

P_B_given_not_A = (B & ~A).mean() / (~A).mean()
print("P(B|not A) =", P_B_given_not_A)

P_A_given_B_bayes = (
    P_B_given_A * P_A
) / (
    (P_B_given_A * P_A) +
    (P_B_given_not_A * P_not_A)
)
print("P(A|B) using Bayes' theorem =", P_A_given_B_bayes)
```

```python
print("P(A|B) direct =", P_A_given_B)
print("P(A|B) using Bayes =", P_A_given_B_bayes)
```

Using Bayes’ theorem, the probability that a student has a part-time job given that the student has a stress score of at least 7 is approximately 68.89%. This agrees exactly with the value obtained using the direct conditional-probability formula, verifying the calculation.

Since P(A ∩ B)=0.05167 is not equal to P(A)P(B)=0.019, the events are not independent. There appears to be an association between having a part-time job and having a high stress score in this dataset.

#Normal Distribution

```python
mean_readiness = df['Academic_Readiness_Score'].mean()
std_readiness = df['Academic_Readiness_Score'].std()

print("Mean =", mean_readiness)
print("Standard Deviation =", std_readiness)
```

```python
highest_score = df['Academic_Readiness_Score'].max()
lowest_score = df['Academic_Readiness_Score'].min()

print("Highest score:", highest_score)
print("Lowest score:", lowest_score)
```

```python
z_highest = (highest_score - mean_readiness) / std_readiness

print("Z-score of highest score:", z_highest)
```

Student's Academic Readiness Score is 2.705 standard deviations above the mean.

```python
z_lowest = (lowest_score - mean_readiness) / std_readiness

print("Z-score of lowest score:", z_lowest)
```

Student's Academic Readiness Score is 3.08 standard deviations below the mean.

##68-95-99.7 Rule
Within 1 standard deviation → approximately 68%

Within 2 standard deviations → approximately 95%

Within 3 standard deviations → approximately 99.7%

```python
print("Within 1 SD:", mean_readiness - std_readiness, "to", mean_readiness + std_readiness)

print("\nWithin 2 SD:", mean_readiness - 2*std_readiness, "to", mean_readiness + 2*std_readiness)

print("\nWithin 3 SD:", mean_readiness - 3*std_readiness, "to", mean_readiness + 3*std_readiness)
```
