#Cleaning a messy company employee dataset

```python
import pandas as pd
```

```python
df = pd.read_csv(r"/content/drive/MyDrive/Colab Notebooks/IDRA/Day11_Messy_Company_Employee_Dataset.csv")
```

```python
df = pd.DataFrame(df)
df.head()
```

```python
df.info()
```

There are 157 rows. It can be seen that there are
- 5 null values in the column 'Department'
- 4 null values in the column 'Age'
- 5 null values in the column 'Gender'
- 5 null values in the column 'Annual_Salary'
- 3 null values in the column 'Experience_Years'
- 5 null values in the column 'City'
- 3 null values in the column 'Performance_Score'
- 2 null values in the column 'Work_Mode'

This can be automatically generated using *isnull().sum()*

```python
df.isnull().sum()
```

Total no. of null values can be seen using *isnull().sum().sum()*

```python
df.isnull().sum().sum()
```

#Dealing with the null values

```python
df['Department'] = df['Department'].fillna(df['Department'].mode()[0])
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])
df['Annual_Salary'] = df['Annual_Salary'].fillna(df['Annual_Salary'].median())
df['Experience_Years'] = df['Experience_Years'].fillna(df['Experience_Years'].median())
df['Performance_Score'] = df['Performance_Score'].fillna(df['Performance_Score'].median())
df['City'] = df['City'].fillna(df['City'].mode()[0])
df['Work_Mode'] = df['Work_Mode'].fillna(df['Work_Mode'].mode()[0])
```

#Inconsistent entries

```python
print(df['Department'].unique())
print(df['Gender'].unique())
print(df['City'].unique())
print(df['Work_Mode'].unique())
```

```python
df['City'] = df['City'].str.strip()
print(df['City'].unique())
```

```python
df['Department'] = df['Department'].str.title()
df['Gender'] = df['Gender'].str.title()
df['City'] = df['City'].str.title()
df['Work_Mode'] = df['Work_Mode'].str.title()

print(df['Department'].unique())
print(df['Gender'].unique())
print(df['City'].unique())
print(df['Work_Mode'].unique())
```

```python
print(df.dtypes)
```

```python
df['Age'] = pd.to_numeric(df['Age'])
df['Experience_Years'] = pd.to_numeric(df['Experience_Years'])
df['Annual_Salary'] = pd.to_numeric(df['Annual_Salary'])
df['Performance_Score'] = pd.to_numeric(df['Performance_Score'])
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'])
```
