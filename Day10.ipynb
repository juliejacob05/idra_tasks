#Data transformation

```python
def occ_rate(row):
  return (row['Passengers']/row['Seat_Capacity'])*100

df['Occupancy_Rate'] = df.apply(occ_rate, axis=1)
df.head()
```

```python
def avail_seats(row):
  return row['Seat_Capacity'] - row['Passengers']

df['Available_Seats'] = df.apply(avail_seats, axis=1)
df.head()
```

There are negative values, which could indicate there are people in the waiting list

```python
def delay_status(row):
  x = row['Delay_Minutes']
  if x == 0:
    return 'On Time'
  elif x>0 and x<=10:
    return 'Slight Delay'
  elif x>10 and x<=30:
    return 'Moderate Delay'
  else:
    return 'Severe Delay'

df['Delay_Condition'] = df.apply(delay_status, axis=1)
df.head()
```

#Airline comparison based on passengers, delays and customer satisfaction

```python
grp = df.groupby('Airline')
summary = grp.agg(
    Num_Flights = ('Flight_ID', 'count'),
    Total_Passengers = ('Passengers', 'sum'),
    Avg_Delay = ('Delay_Minutes', 'mean'),
    Avg_Satisfaction = ('Passenger_Satisfaction', 'mean')
)
print(summary)
```

From the summary table,
- most used airline = Vistara
- most delayed airline = IndiGo
- least dealyed airline = Air India Express
- most satisfactory = Air India
