#Overall Sales Statistics

```python
print("Total sales: ", df['Total_Sales'].sum())
print("Average sales: ", df['Total_Sales'].mean())
print("Total no. of orders: ", df['Order_ID'].count())
print("Total no. of quantities sold: ", df['Quantity'].sum())
```

```python
max_sales = df['Total_Sales'].max()
print("\nHighest sales: ", max_sales, " by")
df[df['Total_Sales']==max_sales]
```

```python
min_sales = df['Total_Sales'].min()
print("\nLowest sales: ", min_sales, " by")
df[df['Total_Sales']==min_sales]
```

After analysing, we can see that
- running sports (5) bought by Ishita Gupta is the highest individual sales in this dataset
- book The Alchemist bought by 2 customers Manya Rao and Aman Kumar are the smallest individual sales

#Groupby

##Category wise Analysis

```python
cat_total_sales = df.groupby('Category')['Total_Sales']
print("Total sales:\n", cat_total_sales.sum())
print("\nAverage sales:\n", cat_total_sales.mean())
print("\nBest performing catgory: ", cat_total_sales.sum().idxmax())
print("Worst performing catgory: ", cat_total_sales.sum().idxmin())
```

```python
cat_quan = df.groupby('Category')['Quantity']
print("Total quantities: \n", cat_quan.sum())

cat_ord = df.groupby('Category')['Order_ID'].count()
print("\n\nNo. of orders per category:\n", cat_ord)
print("\nCategory with highest no. of orders:\n", cat_ord[cat_ord==cat_ord.max()])
```

According to categories, we can see that
- Sports is the best performing category
- also Sports is the category with highest quantity sold

##City wise analysis

```python
city_total_sales = df.groupby('City')['Total_Sales'].sum()
print("Total sales per city:\n", city_total_sales)
print("\nCity with highest total sales: ", city_total_sales.idxmax())
print("City with lowest total sales: ", city_total_sales.idxmin())
```

```python
city_quan = df.groupby('City')['Quantity'].sum()
print("Total quantities per city: ", city_quan)
print("\nCity with highest quantity sold: ", city_quan.idxmax())

city_ord = df.groupby('City')['Order_ID'].count()
print("\n\nNo. of orders per city: ", city_ord)
print("\nCity with highest no. of orders:\n", city_ord[city_ord==city_ord.max()])
```

Grouping by city,
- Hyderabad has the highest total sales
- Pune has most no. of quantities sold
- the highest no. of orders come from Chennai and Pune

##Product wise analysis

```python
prod_total_sales = df.groupby('Product')['Total_Sales'].sum()
print("Total sales per product:\n", prod_total_sales)
print("\nTop-selling product: ", prod_total_sales.idxmax())
print("Least-selling product: ", prod_total_sales.idxmin())
```

```python
prod_no = df.groupby('Product')['Quantity'].sum()
print("Total quantities per product:\n", prod_no)
print("\nProduct with highest quantity sold: ", prod_no.idxmax(), "(", prod_no.max(), ")")
print("Product with lowest quantity sold: ", prod_no.idxmin(), "(", prod_no.min(), ")")
```

```python
prod_ord = df.groupby('Product')['Order_ID'].count()
print("Total orders per product:\n", prod_ord)
print("\nProduct with highest no. of orders: ", prod_ord.idxmax(), "(", prod_ord.max(), ")")
print("Product with lowest no. of orders: ", prod_ord.idxmin(), "(", prod_ord.min(), ")")
```

We can see that
- Coffee Maker is the top selling product
- Wireless headHeadphones had the most no. of quantities sold
- Largest no. of orders come in for Coffee Maker

##Payment Method

```python
pay = df.groupby('Payment_Method')['Total_Sales'].sum()
print("Total sales per payment method:\n", pay)
print("\nPayment method with highest sales: ", pay.idxmax())
print("Payment method with lowest sales: ", pay.idxmin())
```

```python
pay_ord = df.groupby('Payment_Method')['Order_ID'].count()
print("Total orders per payment method:\n", pay_ord)
print("\nPayment method with highest no. of orders: ", pay_ord.idxmax())
print("Payment method with lowest no. of orders: ", pay_ord.idxmin())
```

We can see that
- Most sales are paid using credit card
