#Missing values

```python
df.isna().sum()
```

```python
df.isnull().sum()
```

- **isnull()** is alias for isna()

#**Recap**
- read_csv function for loading the dataset in .csv format
- head() to show the first 5 rows
- tail() to show the last 5 rows
- sample() with no_of_rows argument to show random rows
- .shape gives the no. of rows and columns
- .columns gives the name of the columns in the DataFrame
- .index gives the indices of the DataFrame
- .dtype gives the datatype of each column in the DataFrame
- .info() gives the overall structure of the DataFrame
- .describe() gives the descriptive statistics of the DataFrame
- isna() and isnull() checks if there are any missing values in each column
