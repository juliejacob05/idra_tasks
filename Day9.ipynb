```python
import pandas as pd
import numpy as np
import datetime as dt
```

```python
ord_df = pd.read_csv(r'/content/drive/MyDrive/Colab Notebooks/IDRA/Day9_Orders.csv')
cust_df = pd.read_csv(r'/content/drive/MyDrive/Colab Notebooks/IDRA/Day9_Customers.csv')
prod_df = pd.read_csv(r'/content/drive/MyDrive/Colab Notebooks/IDRA/Day9_Products.csv')
```

#Merging datasets

```python
ord_cust = pd.merge(ord_df, cust_df, on='Customer_ID')
ord_cust.head()
```

```python
ord_cust_prod = pd.merge(ord_cust, prod_df, on='Product_ID')
ord_cust_prod.head()
```

###Removing unnecessary columns

```python
ord_cust_prod.drop(['Customer_ID', 'Product_ID'], axis=1, inplace=True)
ord_cust_prod
```

```python
df = ord_cust_prod[['Order_ID', 'Order_Date', 'Customer_Name', 'City', 'Region', 'Category', 'Product_Name', 'Quantity', 'Unit_Price', 'Brand', 'Payment_Method', 'Order_Status', 'Membership_Type']]
df.info()
```

#DateTime processing

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df.info()
```

```python
df['Order_Date'] = df['Order_Date'].dt.strftime('%d-%m-%Y')
df.head(3)
```

```python
x = df[df['Order_ID']=='O0001']['Order_Date']
x = pd.to_datetime(x, format='%d-%m-%Y')
print(x)

print('\nDateTime Info of the Order_ID O0001:')
print("Date = ", x.dt.day.values[0])
print("Month = ", x.dt.month.values[0])
print("Year = ", x.dt.year.values[0])

print("\nDerived Info:")
print("Day = ", x.dt.day_name().values[0])
print("Day of the week = ", x.dt.day_of_week.values[0])
print("Month name = ", x.dt.month_name().values[0])
print("Quarter = ", x.dt.quarter.values[0])
print("Day of the year = ", x.dt.dayofyear.values[0])
print("Week of the year = ", x.dt.isocalendar().week.values[0])
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

#Adding new column, Total_Amount

```python
def tot_amt(row):
  return row['Quantity'] * row['Unit_Price']

df['Total_Amount'] = df.apply(tot_amt, axis=1)
df = df[['Order_ID', 'Order_Date', 'Customer_Name', 'City', 'Region', 'Category', 'Product_Name', 'Quantity', 'Unit_Price', 'Total_Amount', 'Brand', 'Payment_Method', 'Order_Status', 'Membership_Type']]
df
```
